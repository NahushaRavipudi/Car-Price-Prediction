```python
#categoricals to change numericals (categorical encoding)
data_set.replace({'Fuel_Type':{'Petrol':0,'Diesel':1,'CNG':2}},inplace=True)
data_set.replace({'Seller_Type':{'Dealer':0,'Individual':1}},inplace=True)
data_set.replace({'Transmission':{'Manual':0,'Automatic':1}},inplace=True)
```

```python
data_set.head()
```

```python
data_set.info()
```

```python
# 3rd process - Splitting data  x,y

x=data_set.drop(['Car_Name','Selling_Price'],axis=1)  #axis=1 means sending them to x axis
y=data_set['Selling_Price']
```

```python
x.head()
```

```python
y.head()
```

```python
#Train test Split

from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
print(data_set.shape , x.shape , y.shape , x_train.shape , y_train.shape , x_test.shape , y_test.shape)
```

```python
#model training
from sklearn.linear_model import LinearRegression
model=LinearRegression()
lr=model.fit(x_train,y_train)
```

```python
#accuracy score for training
# Check how well the Linear Regression model performs on training data
lr.score(x_train,y_train)
```

```python
#accuracy score for testing
# Check how well the Linear Regression model performs on testing data
lr.score(x_test,y_test)
```

```python
# 5th process - predictive system
input_data = (2014, 5.59, 27000, 0, 0, 0, 0)
input_data_as_numpy_array = np.asarray(input_data)
input_data_reshaped = input_data_as_numpy_array.reshape(1, -1)
prediction = lr.predict(input_data_reshaped)
print("prediction price", prediction, "lakhs")
```
